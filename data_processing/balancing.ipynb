# Data Balancing

```python
import pandas as pd

from imblearn.over_sampling import SMOTENC
```

```python
seed = 42
```

```python
dataset = pd.read_csv("../datasets/thyroid0387_imputed.data", sep = ";", index_col="Unnamed: 0")
```

```python
dataset.head()
```

```python
dataset.dropna()["diagnosis_group"].value_counts()
```

The data is heavely unbalanced, let's fix it.

```python
x = dataset.drop(columns = ["diagnosis_group", "diagnosis"])
y = dataset["diagnosis_group"]
```

```python
numeric_col = x.select_dtypes(include = ["int64", "float64"]).columns
categoric_col = x.select_dtypes(exclude = ["int64", "float64"]).columns
categoric_col_indices = [x.columns.get_loc(col) for col in categoric_col]
```

```python
for col in categoric_col:
    x[col] = x[col].astype('object')
```

```python
smote = SMOTENC(random_state = seed, categorical_features=categoric_col_indices)
x_value, y_value = smote.fit_resample(x, y)
```

```python
x_value.isnull().sum()
```

```python
temp = pd.concat([x_value, y_value], axis=1)
```

```python
temp["diagnosis_group"].value_counts()
```

```python
temp.isnull().sum()
```

```python
temp.to_csv("../datasets/thyroid0387_balanced.data", sep=";")
```
